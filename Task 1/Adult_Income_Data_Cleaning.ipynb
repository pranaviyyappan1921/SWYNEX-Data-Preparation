```python
df = pd.read_csv("/content/adult.data")

df.head()
```

```python
# Check the number of rows and columns
print("Rows and columns:", df.shape)

# View column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()

# Check the first five rows
display(df.head())
```

```python
#Step 4: Identify missing values
print(df.isnull().sum())
```

```python
# Count '?' values in each column
for col in df.select_dtypes(include="object").columns:
    count = (df[col].astype(str).str.strip() == "?").sum()
    if count > 0:
        print(col, ":", count, "unknown values")
```

```python
#Step 5: Check duplicate rows
print("Total duplicate rows:", df.duplicated().sum())
```

```python
#Step 6: Check data types
df.info()
```

```python
#Step 7: Check inconsistent values
for col in df.select_dtypes(include="object").columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts().head(10))
```

```python
#Step 8: Record the original dataset statistics
print("Original rows:", df.shape[0])
print("Original columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
#Step 9: Handle missing values represented by ?
# Count '?' values in each column
for col in df.select_dtypes(include="object").columns:
    count = (df[col].astype(str).str.strip() == "?").sum()

    if count > 0:
        print(f"{col}: {count} missing values")
```

```python
#Step 10: Remove extra spaces
# Remove leading and trailing spaces from text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

print("Extra spaces removed.")
```

```python
#Step 11: Convert ? into proper missing values
# Replace '?' with NaN
df = df.replace(r'^\s*\?\s*$', np.nan, regex=True)

# Check missing values again
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
#Step 12: Check missing values again
# Count missing values in each column
print(df.isnull().sum())

# Total missing values
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
#Step 13: Handle missing values
# Find categorical columns
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns

# Fill missing categorical values with the mode
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col] = df[col].fillna(mode_value)

print("Missing values after imputation:",
      df.isnull().sum().sum())
```

```python
#Step 14: Inspect and remove duplicate rows
print("Duplicates before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
print("Dataset shape after removing duplicates:", df.shape)
```

```python
#Step 15: Validate the cleaned dataset
print("Final rows:", df.shape[0])
print("Final columns:", df.shape[1])
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

display(df.head())
```

```python
print("CLEANED DATASET REPORT")
print("----------------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```
