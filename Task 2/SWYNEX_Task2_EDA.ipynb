```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/cleaned_adult_income.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
#checking dataset shape, missing values, and duplicate rows.
# 1. Check dataset shape
print("Dataset Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# 2. Check missing values
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

# 3. Check duplicate rows
print("\nTotal Duplicate Rows:", df.duplicated().sum())
```

```python
import pandas as pd

column_names = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education-num",
    "marital-status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital-gain",
    "capital-loss",
    "hours-per-week",
    "native-country",
    "income"
]

df = pd.read_csv(
    "/content/cleaned_adult_income.csv",
    header=None,
    names=column_names,
    skipinitialspace=True
)

print("Dataset shape:", df.shape)
print(df.columns.tolist())
display(df.head())
```

```python
#Analyze income
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="income")

plt.title("Income Category Distribution")
plt.xlabel("Income Category")
plt.ylabel("Number of People")

plt.show()
```

```python
#Analyze age
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=30, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#Income by gender
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="sex", hue="income")
plt.title("Income Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of People")
plt.show()
```

```python
#Weekly working hours
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="hours-per-week", bins=30, kde=True)
plt.title("Weekly Working Hours Distribution")
plt.xlabel("Hours Worked Per Week")
plt.ylabel("Number of People")
plt.show()
```

```python
#Education and income
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x="education", hue="income")
plt.title("Income Distribution by Education")
plt.xlabel("Education")
plt.ylabel("Number of People")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Calculate summary statistics
print("Dataset shape:", df.shape)

print("\nNumerical Summary:")
display(df.describe())

print("\nAverage age by income:")
display(df.groupby("income")["age"].mean())

print("\nAverage weekly hours by income:")
display(df.groupby("income")["hours-per-week"].mean())

print("\nIncome percentages:")
display((df["income"].value_counts(normalize=True) * 100).round(2))
```

## Five Key Insights from EDA

### 1. Income Distribution
The dataset contains more people earning `<=50K` than people earning `>50K`. The income classes are imbalanced, which should be considered when training a machine-learning model.

### 2. Age Distribution
The age histogram helps us understand the age distribution of individuals in the dataset and identify the age groups that occur most frequently.

### 3. Gender and Income
The comparison of gender and income shows how income categories are distributed across genders. This can help identify patterns and potential bias in income prediction.

### 4. Weekly Working Hours
The weekly working-hours graph shows how many hours people typically work. Comparing average working hours across income groups can help identify differences in work patterns.

### 5. Education and Income
The education and income graph helps us examine how income categories vary across education levels. Education may be a useful feature for predicting income.

## Conclusion
EDA helps us understand the dataset, identify patterns, and select potentially useful features for future machine-learning models.
